```python
from docx import Document
from docx.shared import Pt
from docx.oxml import OxmlElement
from docx.oxml.ns import qn

def set_rtl(paragraph):
    p = paragraph._element
    pPr = p.get_or_add_pPr()
    bidi = OxmlElement('w:bidi')
    bidi.set(qn('w:val'), '1')
    pPr.append(bidi)

def set_font(run, font_name, font_size=14):
    run.font.name = font_name
    run.font.size = Pt(font_size)

    r = run._element
    rPr = r.get_or_add_rPr()
    rFonts = OxmlElement('w:rFonts')
    rFonts.set(qn('w:ascii'), font_name)
    rFonts.set(qn('w:hAnsi'), font_name)
    rFonts.set(qn('w:cs'), font_name)
    rFonts.set(qn('w:eastAsia'), font_name)
    rPr.append(rFonts)

result = ocr_fa.ocr("test_002.jpg")


document = Document()

for text in fix_display(texts):
    p = document.add_paragraph()
    set_rtl(p)
    run = p.add_run(text)
    set_font(run, "B Nazanin", 14)

document.save("ocr_output.docx")
print("فایل Word ساخته شد.")
```

```python
import cv2
import re
import unicodedata
from paddleocr import PaddleOCR
from docx import Document
from docx.shared import Pt
from docx.oxml import OxmlElement
from docx.oxml.ns import qn


# ---------------------------------------------------------
# ۱. دیکشنری تبدیل کاراکترهای Presentation Forms به حروف استاندارد فارسی
# ---------------------------------------------------------
PRESENTATION_FORMS_MAP = {
    'ﺁ': 'آ', 'ﺂ': 'آ', 'ﺃ': 'أ', 'ﺄ': 'أ', 'ﺇ': 'إ', 'ﺈ': 'إ',
    'ﺍ': 'ا', 'ﺎ': 'ا', 'ﺏ': 'ب', 'ﺐ': 'ب', 'ﺑ': 'ب', 'ﺒ': 'ب',
    'ﭖ': 'پ', 'ﭗ': 'پ', 'ﭘ': 'پ', 'ﭙ': 'پ', 'ﺕ': 'ت', 'ﺖ': 'ت',
    'ﺗ': 'ت', 'ﺘ': 'ت', 'ﺙ': 'ث', 'ﺚ': 'ث', 'ﺛ': 'ث', 'ﺜ': 'ث',
    'ﺝ': 'ج', 'ﺞ': 'ج', 'ﺟ': 'ج', 'ﺠ': 'ج', 'ﭺ': 'چ', 'ﭻ': 'چ',
    'ﭼ': 'چ', 'ﭽ': 'چ', 'ﺡ': 'ح', 'ﺢ': 'ح', 'ﺣ': 'ح', 'ﺤ': 'ح',
    'ﺥ': 'خ', 'ﺦ': 'خ', 'ﺧ': 'خ', 'ﺨ': 'خ', 'ﺩ': 'د', 'ﺪ': 'د',
    'ﺫ': 'ذ', 'ﺬ': 'ذ', 'ﺭ': 'ر', 'ﺮ': 'ر', 'ﺯ': 'ز', 'ﺰ': 'ز',
    'ﮊ': 'ژ', 'ﮋ': 'ژ', 'ﺱ': 'س', 'ﺲ': 'س', 'ﺳ': 'س', 'ﺴ': 'س',
    'ﺵ': 'ش', 'ﺶ': 'ش', 'ﺷ': 'ش', 'ﺸ': 'ش', 'ﺹ': 'ص', 'ﺺ': 'ص',
    'ﺻ': 'ص', 'ﺼ': 'ص', 'ﺽ': 'ض', 'ﺾ': 'ض', 'ﺿ': 'ض', 'ﻀ': 'ض',
    'ﻁ': 'ط', 'ﻂ': 'ط', 'ﻃ': 'ط', 'ﻄ': 'ط', 'ﻅ': 'ظ', 'ﻆ': 'ظ',
    'ﻇ': 'ظ', 'ﻈ': 'ظ', 'ﻉ': 'ع', 'ﻊ': 'ع', 'ﻋ': 'ع', 'ﻌ': 'ع',
    'ﻍ': 'غ', 'ﻎ': 'غ', 'ﻏ': 'غ', 'ﻐ': 'غ', 'ﻑ': 'ف', 'ﻒ': 'ف',
    'ﻓ': 'ف', 'ﻔ': 'ف', 'ﻕ': 'ق', 'ﻖ': 'ق', 'ﻗ': 'ق', 'ﻘ': 'ق',
    'ﻙ': 'ک', 'ﻚ': 'ک', 'ﻛ': 'ک', 'ﻜ': 'ک', 'ﮎ': 'ک', 'ﮏ': 'ک',
    'ﮐ': 'ک', 'ﮑ': 'ک', 'ﮒ': 'گ', 'ﮓ': 'گ', 'ﮔ': 'گ', 'ﮕ': 'گ',
    'ﻝ': 'ل', 'ﻞ': 'ل', 'ﻟ': 'ل', 'ﻠ': 'ل', 'ﻡ': 'م', 'ﻢ': 'م',
    'ﻣ': 'م', 'ﻤ': 'م', 'ﻥ': 'ن', 'ﻦ': 'ن', 'ﻧ': 'ن', 'ﻨ': 'ن',
    'ﻭ': 'و', 'ﻮ': 'و', 'ﻩ': 'ه', 'ﻪ': 'ه', 'ﻫ': 'ه', 'ﻬ': 'ه',
    'ﯼ': 'ی', 'ﯽ': 'ی', 'ﯾ': 'ی', 'ﯿ': 'ی', 'ﻱ': 'ی', 'ﻲ': 'ی',
    'ﻳ': 'ی', 'ﻴ': 'ی', 'ﺋ': 'ئ', 'ﺌ': 'ئ', 'ﺀ': 'ء'
}

def fix_persian_text(text: str) -> str:
    """اصلاح حروف چسبیده و معکوس شده فارسی"""
    if not text:
        return ""
    
    # ۱. تبدیل کاراکترهای غیراستاندارد به فارسی اصلی
    res = []
    for char in text:
        res.append(PRESENTATION_FORMS_MAP.get(char, char))
    corrected = "".join(res)
    
    # ۲. اگر متن حاوی حروف فارسی باشد و برعکس شده باشد، آن را معکوس می‌کنیم
    if re.search(r'[\u0600-\u06FF]', corrected):
        # معکوس کردن رشته برای درست شدن ترتیب حروف
        corrected = corrected[::-1]
        
    return corrected.strip()

def normalize_digits(text: str) -> str:
    """تبدیل ارقام فارسی/عربی به انگلیسی و اصلاح اعداد برعکس شده"""
    if not text:
        return ""
    
    fa_digits = "۰۱۲۳۴۵۶۷۸۹"
    ar_digits = "٠١٢٣٤٥٦٧٨٩"
    en_digits = "0123456789"
    
    # تبدیل به ارقام انگلیسی برای یکدست‌سازی
    for f, a, e in zip(fa_digits, ar_digits, en_digits):
        text = text.replace(f, e).replace(a, e)
        
    # اگر کلمه فقط عدد است و برعکس شده (مثل 9831 برای 1389)
    if text.isdigit() and len(text) > 1:
        # برای سال تولد یا مدل‌های ۴ رقمی (مثلاً ۱۳۸۹)
        if len(text) == 4 and text.startswith(('9', '8')): 
            text = text[::-1]
            
    return text


# ---------------------------------------------------------
# ۲. کانفیگ اصلاح‌شده با مختصات‌های دقیق‌تر
# ---------------------------------------------------------
CONFIG = {
    "name":               {"roi": (0.3273, 0.367, 0.5407, 0.8975), "lang": "fa", "type": "text"},
    "name_father":        {"roi": (0.3205, 0.3541, 0.2963, 0.4165), "lang": "fa", "type": "text"},
    "تاریخ_تولد":         {"roi": (0.3522, 0.4098, 0.6598, 0.7943), "lang": "fa", "type": "date"},
    "شماره شناسنامه":     {"roi": (0.3938, 0.4374, 0.6733, 0.7567), "lang": "fa", "type": "numeric"},
    "شماره ملی":          {"roi": (0.3889, 0.4346, 0.2694, 0.3924), "lang": "fa", "type": "numeric"},
    "شماره پلاک":         {"roi": (0.4286, 0.4862, 0.1357, 0.3382), "lang": "fa", "type": "plate"},
    "کدپستی":             {"roi": (0.4294, 0.4702, 0.6574, 0.805), "lang": "fa", "type": "numeric"},
    "Addres":             {"roi": (0.4722, 0.503, 0.1701, 0.8892), "lang": "fa", "type": "text"},
    "نوع":                {"roi": (0.2084, 0.2381, 0.7472, 0.8975), "lang": "fa", "type": "text"},
    "سیستم":              {"roi": (0.2073, 0.2332, 0.5479, 0.7108), "lang": "fa", "type": "text"},
    "تیپ":                {"roi": (0.2104, 0.2352, 0.3821, 0.5479), "lang": "fa", "type": "text"},
    "مدل":                {"roi": (0.2124, 0.2361, 0.2286, 0.354), "lang": "fa", "type": "numeric"},
    "رنگ":                {"roi": (0.2005, 0.2361, 0.0763, 0.2401), "lang": "fa", "type": "text"},  # مختصات رنگ اصلاح شد
    "موتور":              {"roi": (0.2501, 0.2737, 0.5542, 0.856), "lang": "en", "type": "code"},
    "شماره_شاسی":         {"roi": (0.2521, 0.2709, 0.2318, 0.4719), "lang": "en", "type": "code"},
    "VIN":                {"roi": (0.2689, 0.3017, 0.2413, 0.7619), "lang": "en", "type": "code"},  # مختصات VIN اصلاح شد
    "شماره_سند":          {"roi": (0.0208, 0.0596, 0.0763, 0.2536), "lang": "fa", "type": "numeric"},
    "شماره برگ سبز":      {"roi": (0.3174, 0.4622, 0.0669, 0.123), "lang": "fa", "type": "numeric"}
}

# ---------------------------------------------------------
# ۳. تابع اصلی استخراج با مرتب‌سازی فضایی RTL
# ---------------------------------------------------------
ocr_fa = PaddleOCR(use_angle_cls=True, lang="fa", show_log=False)
ocr_en = PaddleOCR(use_angle_cls=True, lang="en", show_log=False)

def process_document_roi(image_path):
    img = cv2.imread(image_path)
    if img is None:
        raise FileNotFoundError("تصویر پیدا نشد!")
        
    h, w, _ = img.shape
    final_results = {}

    for field, info in CONFIG.items():
        roi = info["roi"]
        lang = info.get("lang", "fa")
        field_type = info.get("type", "text")
        
        # تبدیل درصد به پیکسل
        y1, y2 = int(roi[0] * h), int(roi[1] * h)
        x1, x2 = int(roi[2] * w), int(roi[3] * w)

        crop = img[y1:y2, x1:x2]
        if crop.size == 0:
            continue

        # اجرای OCR
        ocr_engine = ocr_fa if lang == "fa" else ocr_en
        res = ocr_engine.ocr(crop, cls=True)

        extracted_words = []
        if res and res[0]:
            # مرتب‌سازی کلمات از راست به چپ بر اساس مختصات X کادرها (برای زبان فارسی)
            lines = res[0]
            if lang == "fa":
                # کادرها براساس x_max معکوس مرتب می‌شوند (از راست به چپ)
                lines = sorted(lines, key=lambda b: b[0][0][0], reverse=True)
            
            for line in lines:
                raw_text = line[1][0]
                extracted_words.append(raw_text)

        raw_joined = " ".join(extracted_words)

        # مرحله پاک‌سازی بر اساس نوع فیلد
        if lang == "fa":
            cleaned_text = fix_persian_text(raw_joined)
        else:
            cleaned_text = raw_joined.strip()

        cleaned_text = normalize_digits(cleaned_text)

        # فیلترهای اختصاصی بر اساس نوع داده
        if field_type == "numeric":
            cleaned_text = re.sub(r'[^\d]', '', cleaned_text)
        elif field_type == "date":
            # استخراج الگوی تاریخ YYYY/MM/DD
            dates = re.findall(r'\d{4}/\d{1,2}/\d{1,2}', cleaned_text)
            if dates:
                cleaned_text = dates[0]

        final_results[field] = cleaned_text
        print(f"[{field:15}] -> {cleaned_text}")

    return final_results


def set_rtl(paragraph):
    p = paragraph._element
    pPr = p.get_or_add_pPr()
    bidi = OxmlElement('w:bidi')
    bidi.set(qn('w:val'), '1')
    pPr.append(bidi)

def set_font(run, font_name, font_size=14):
    run.font.name = font_name
    run.font.size = Pt(font_size)

    r = run._element
    rPr = r.get_or_add_rPr()
    rFonts = OxmlElement('w:rFonts')
    rFonts.set(qn('w:ascii'), font_name)
    rFonts.set(qn('w:hAnsi'), font_name)
    rFonts.set(qn('w:cs'), font_name)
    rFonts.set(qn('w:eastAsia'), font_name)
    rPr.append(rFonts)

# اجرای تابع
results = process_document_roi("test_002.jpg")
document = Document()

for text in results:
    p = document.add_paragraph()
    set_rtl(p)
    run = p.add_run(text)
    set_font(run, "B Nazanin", 14)

document.save("ocr_output.docx")
print("فایل Word ساخته شد.")
```
